## Atividade 02

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Analisando os dados que serão usados nos gráficos
print('Quantidade de alunos no df: ', df['id_aluno'].nunique())
print('Quantidade de status no df: ', df['status'].nunique())
print('Status que temos no df: ', df['status'].unique())

# 2. Preparando os dados com groupby
contagem_status = df.groupby('status')['id_aluno'].nunique().reset_index()
contagem_status.columns = ['status', 'quantidade']
contagem_status = contagem_status.sort_values('quantidade', ascending=False)

cores = sns.color_palette('pastel', n_colors=len(contagem_status))

# 3. Montando o dashboard (1 linha, 2 colunas)
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# 4. ---- Gráfico 1: Barras ----
ax1 = axes[0]
sns.barplot(
    data=contagem_status,
    x='status',
    y='quantidade',
    hue='status',
    palette=cores,
    legend=False,
    ax=ax1
)
for container in ax1.containers:
    ax1.bar_label(container, padding=3)

ax1.set_title('Quantidade de alunos por status', fontsize=13)
ax1.set_xlabel('Status')
ax1.set_ylabel('Quantidade de alunos')
ax1.spines[['top', 'right']].set_visible(False)

# 5. ---- Gráfico 2: Pizza ----
ax2 = axes[1]
ax2.pie(
    contagem_status['quantidade'],
    labels=contagem_status['status'],
    autopct='%1.1f%%',
    startangle=90,
    colors=cores,
    wedgeprops={'edgecolor': 'white'}
)
ax2.set_title('Percentual de alunos por status', fontsize=13)
ax2.axis('equal')

# 6. ---- Dashboard Geral ----
fig.suptitle(
    'Dashboard - Alunos por Status',
    fontsize=16,
    fontweight='bold')

plt.tight_layout()
plt.show()
```
